# 利用シーン
## シーン1：ICAOコードで検索したい
ICAOコードで飛行場を検索するなら、`search_icao_airport_code`を使います。

#### search_icao_airport_code
`search_icao_airport_code`は指定したICAOコードの飛行場情報を取得します。

```python
from nakametpy.aviation import airport_info
from pprint import pprint

airport = airport_info()
pprint(airport.search_icao_airport_code("RJAA"), width=100)
```
