# 利用シーン
## シーン1：Bufrの資料記述子を取得したい
第3節の資料記述子を確認するなら、`get_data_description`または`get_data_descriptors`を使います。

#### get_data_description
`get_data_description`は資料記述子そのものの説明のみを出力します。

```python
from nakametpy.bufr import bufr
from pprint import pprint

data_path = "../tests/data/bufr/bufr/IUKL01_RJTD_300000_202410300123114_001.send" # 東京編集の地上高層実況気象報
bufr_class = bufr(data_path)
pprint(bufr_class.get_data_description(), width=90)
```

#### get_data_descriptors
`get_data_descriptors`は資料記述子の詳細な説明を出力します。集約記述子の場合は、集約された記述子についても出力します。

```python
from nakametpy.bufr import bufr
from pprint import pprint

data_path = "../tests/data/bufr/bufr/IUKL01_RJTD_300000_202410300123114_001.send" # 東京編集の地上高層実況気象報
bufr_class = bufr(data_path)
pprint(bufr_class.get_data_descriptors(), width=90)
```

#### get_extracted_data_description
`get_extracted_data_description`は集約記述子を展開した資料記述子の説明を出力します。

```python
from nakametpy.bufr import bufr
from pprint import pprint

data_path = "../tests/data/bufr/bufr/IUKL01_RJTD_300000_202410300123114_001.send" # 東京編集の地上高層実況気象報
bufr_class = bufr(data_path)
pprint(bufr_class.get_extracted_data_description(), width=90)
```

#### get_extracted_data_descriptors
`get_extracted_data_descriptors`は集約記述子を展開した資料記述子の説明を出力します。またその要素の深さの情報も合わせて出力されます。

```python
from nakametpy.bufr import bufr
from pprint import pprint

data_path = "../tests/data/bufr/bufr/IUKL01_RJTD_300000_202410300123114_001.send" # 東京編集の地上高層実況気象報
bufr_class = bufr(data_path)
pprint(bufr_class.get_extracted_data_descriptors(), width=100)
```

#### read_data
`read_data`はデータを読み込みます。

```python
from nakametpy.bufr import bufr
from pprint import pprint

data_path = "../tests/data/bufr/bufr/IUKL01_RJTD_300000_202410300123114_001.send" # 東京編集の地上高層実況気象報
bufr_class = bufr(data_path)
data = bufr_class.read_data()
pprint(data)
```
